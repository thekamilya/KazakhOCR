```python
import re
import random

# Regular expression for Kazakh/Cyrillic letters
letters_only_pattern = re.compile(r'^[а-яА-ЯёЁәӘғҒқҚңҢөӨұҰүҮһҺіІ]+$')
all_phrases = set()

for sentence in ds['train']['kk']:
    # Stop everything if we've reached our 4000 sample limit
    if len(all_phrases) >= 4000:
        break

    if not isinstance(sentence, str):
        continue

    # 1. Extract only valid words
    raw_words = sentence.split()
    valid_words = []

    for word in raw_words:
        clean_word = word.strip('.,!?:;"()»«-')
        if len(clean_word) >= 4 and letters_only_pattern.match(clean_word):
            valid_words.append(clean_word)

    # 2. Process only if we have at least 3 words
    if len(valid_words) >= 3:
        # Determine random length (3, 4, or 5)
        n = random.randint(3, 5)

        # Ensure n isn't larger than the actual word count available
        n = min(n, len(valid_words))

        # Pick a random starting point for the phrase
        start_index = random.randint(0, len(valid_words) - n)

        phrase = " ".join(valid_words[start_index : start_index + n])
        print(phrase)
        all_phrases.add(phrase)

print(f"Extraction complete. Total unique samples: {len(all_phrases)}")
```

```python
import os

font_dir = "./fonts"

FONT_PATHS = [
    os.path.join(font_dir, f)
    for f in os.listdir(font_dir)
    if f.lower().endswith((".ttf", ".otf"))
]

print(FONT_PATHS)
```

```python
import os
import random
import pandas as pd
from PIL import Image, ImageDraw, ImageFont, ImageFilter
import numpy as np


OUTPUT_DIR = "/content/drive/MyDrive/Kazakh_OCR_Dataset/images"
CSV_FILE = "/content/drive/MyDrive/Kazakh_OCR_Dataset/dataset_mapping.csv"
os.makedirs(OUTPUT_DIR, exist_ok=True)


def add_noise(img):
    """Adds random grain/noise to the image."""
    np_img = np.array(img)
    noise = np.random.normal(0, random.uniform(2, 10), np_img.shape).astype('int16')
    np_img = np.clip(np_img.astype('int16') + noise, 0, 255).astype('uint8')
    return Image.fromarray(np_img)

def generate_ocr_image(text, filename):
    # 1. Random Background Color (light vs dark)
    is_dark_mode = random.random() > 0.8
    if is_dark_mode:
        bg_color = (random.randint(0, 50), random.randint(0, 50), random.randint(0, 50))
        text_color = (random.randint(150, 255), random.randint(150, 255), random.randint(150, 255))
    else:
        bg_color = (random.randint(200, 255), random.randint(200, 255), random.randint(200, 255))
        text_color = (random.randint(0, 100), random.randint(0, 100), random.randint(0, 100))

    # 2. Dynamic Image Size
    img_w = len(text) * 40 + 80
    img_h = 120
    image = Image.new('RGB', (img_w, img_h), color=bg_color)
    draw = ImageDraw.Draw(image)

    # 3. Random Font & Size
    try:
        font_path = random.choice(FONT_PATHS)
        font = ImageFont.truetype(font_path, random.randint(40, 60))
    except:
        font = ImageFont.load_default()

    # 4. Center text with random jitter
    bbox = draw.textbbox((0, 0), text, font=font)
    tw, th = bbox[2] - bbox[0], bbox[3] - bbox[1]
    x = (img_w - tw) / 2 + random.randint(-20, 20)
    y = (img_h - th) / 2 + random.randint(-10, 10)
    draw.text((x, y), text, fill=text_color, font=font)

    # 5. Augmentations: Rotation
    image = image.rotate(random.uniform(-5, 5), expand=False, fillcolor=bg_color)

    # 6. Augmentations: Blur (Lower Quality)
    if random.random() > 0.5:
        image = image.filter(ImageFilter.GaussianBlur(radius=random.uniform(0.1, 1.5)))

    # 7. Augmentations: Noise
    image = add_noise(image)

    image.save(os.path.join(OUTPUT_DIR, filename), quality=random.randint(30, 95))

# --- EXECUTION ---
data_log = []

print(f"Generating {len(all_phrases)} images...")
for i, word in enumerate(all_phrases):
    img_name = f"kaz_{i:06d}.jpg"
    generate_ocr_image(word, img_name)
    data_log.append({"filename": img_name, "text": word})

# Save the CSV mapping
df = pd.DataFrame(data_log)
df.to_csv(CSV_FILE, index=False, encoding='utf-8')

print(f"Done! Images in /{OUTPUT_DIR} and labels in {CSV_FILE}")
```
